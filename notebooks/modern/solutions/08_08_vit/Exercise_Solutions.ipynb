## Exercise 2 — Parameter count and latency

```python
import time,torch
def model_stats(model,input_shape=(1,3,224,224),runs=30):
    dev=next(model.parameters()).device; x=torch.randn(*input_shape,device=dev); model.eval()
    with torch.inference_mode():
        for _ in range(10): model(x)
        if dev.type=='cuda': torch.cuda.synchronize()
        t=time.perf_counter()
        for _ in range(runs): model(x)
        if dev.type=='cuda': torch.cuda.synchronize()
    return sum(p.numel() for p in model.parameters()),(time.perf_counter()-t)*1000/runs
# Compare stats on identical hardware and report input shape, batch size and precision.
```
