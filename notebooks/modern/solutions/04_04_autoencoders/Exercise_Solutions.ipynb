## Exercise 3 — Compression ratio
For a dense latent vector, account for latent values, quantization bits, dimensions, model size and side information. A neural autoencoder is not automatically a practical codec.

```python
def raw_bits(height,width,channels=1,bits_per_pixel=8):
    return height*width*channels*bits_per_pixel
def latent_bits(latent_count,bits_per_latent=8,metadata_bits=0):
    return latent_count*bits_per_latent+metadata_bits
original=raw_bits(28,28)
compressed=latent_bits(32,bits_per_latent=8,metadata_bits=64)
print({'original_bits':original,'latent_plus_metadata_bits':compressed,'ratio':original/compressed})
```
